```python
sum = 0 #### sum variable initiated to 0
print("Before the loop the sum is", sum)
for thing in [9, 41, 12, 3, 74, 15] :
 sum= sum+ thing
 print(sum, thing)
print ("After the loop the sum is", sum)
```

###Task 8: Write a nested for loop for the below two list of people and places ensuring that each person goes to each place.

```python
persons = [ "John", "Peter", "Vikki", "Henry" ]
places = [ "Spain", "Italy", "India", "UK" ]
for x in persons:
  for y in places:
    print(x,"visits",y)
```
