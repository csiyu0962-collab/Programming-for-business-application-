#Exercise 3: if...elif...else

##Task 1

```python
amount = int(input("Enter amount: "))
if amount<1000:
 print("Code being executed from 'if' section")
 discount = amount*0.05
 print ("Discount",discount)
elif amount<5000:
 print("Code being executed from 'elif' section")
 discount = amount*0.10
 print ("Discount",discount)
else:
 print("Code being executed from 'else' section")
 discount = amount*0.15
 print ("Discount",discount)

print ("Net payable:",amount-discount)
```
