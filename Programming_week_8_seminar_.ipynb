##Task 1: Basic Summary of the Dataset

```python
import pandas as pd
# Load your dataset
df = pd.read_csv("/content/IMDBMovies.csv")
# Get the basic information about the dataset
data_info = df.info()
```

```python
# Get a summary of the numerical columns (mean, median, etc.)
data_summary = df.describe()
data_summary
```

```python
# Check for missing values in each column
missing_values = df.isnull().sum()
missing_values
```

##Task 2: Check Unique Values for Categorical Columns

```python
# Check unique values for categorical columns
unique_genres = df['Genre'].unique()
unique_certificates = df['Certificate'].unique()
unique_directors = df['Director'].unique()
# Count the number of unique values for each
num_unique_genres = df['Genre'].nunique()
num_unique_certificates = df['Certificate'].nunique()
num_unique_directors = df['Director'].nunique()
print("Unique Genres:", unique_genres)
print("Number of unique Genres:", num_unique_genres)
print("Unique Certificates:", unique_certificates)
print("Number of unique Certificates:", num_unique_certificates)
print("Unique Directors:", unique_directors)
print("Number of unique Directors:", num_unique_directors)
```

##Task 3: Movies with the Highest Gross Earnings

```python
# Convert 'Gross' to numeric, handling any errors (e.g., missing or
#malformed values)
df['Gross'] = pd.to_numeric(df['Gross'], errors='coerce')
# Get the movie with the highest Gross earnings
highest_gross_movie = df.loc[df['Gross'].idxmax()]
print("Highest grossing movie:\n", highest_gross_movie[['Series_Title',
'Gross']])
```
